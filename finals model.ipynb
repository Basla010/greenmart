```python
#K-Medoids
```

```python
def k_medoids(X, n_clusters, random_state=42, max_iter=100):
    rng = np.random.RandomState(random_state)
    n_samples = X.shape[0]
    
    medoid_idx = rng.choice(n_samples, n_clusters, replace=False)
    medoids = X[medoid_idx]
    
    for _ in range(max_iter):
        dist = np.linalg.norm(X[:, None] - medoids[None, :], axis=2)
        labels = dist.argmin(axis=1)
        
        new_medoid_idx = medoid_idx.copy()
        for k in range(n_clusters):
            cluster_points_idx = np.where(labels == k)[0]
            if len(cluster_points_idx) == 0:
                continue
            cluster_points = X[cluster_points_idx]
            intra_dist = np.linalg.norm(cluster_points[:, None] - cluster_points[None, :], axis=2).sum(axis=1)
            new_medoid_idx[k] = cluster_points_idx[intra_dist.argmin()]
        
        if np.array_equal(new_medoid_idx, medoid_idx):
            break
        medoid_idx = new_medoid_idx
        medoids = X[medoid_idx]
    
    return labels, medoids
```

```python
labels_kmed, medoids = k_medoids(x_cluster_scaled, n_clusters=4, random_state=42)

df['Cluster_KMedoids'] = labels_kmed
df['Cluster_KMedoids'].value_counts()
```

```python
cluster_profile_kmed = df.groupby('Cluster_KMedoids')[cluster_features].mean()
print(cluster_profile_kmed)

plt.figure(figsize=(8,6))
sns.scatterplot(x='Income', y='Total_Spend', hue='Cluster_KMedoids', data=df, palette='Set2', alpha=0.7)
plt.title('K-Medoids: Income vs Total Spend')
plt.show()
```
